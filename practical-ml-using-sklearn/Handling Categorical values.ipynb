```python
import pandas as pd
```

```python
df = pd.read_csv("housing.csv")
```

```python
import numpy as np
```

```python
df["income_cat"]=pd.cut(df["median_income"], bins=[0, 1.5, 3.0, 4.5, 6.0, 7.5, np.inf], labels=[1,2,3,4,5,6])
```

```python
from sklearn.model_selection import StratifiedShuffleSplit
split=StratifiedShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
for train_index, test_index in split.split(df, df["income_cat"]):
    strat_train_set = df.loc[train_index]
    strat_test_set = df.loc[test_index]
```

```python
for sett in (strat_train_set, strat_test_set):
    sett.drop(labels="income_cat", inplace=True, axis=1)
```

```python
df =strat_train_set.copy()
```

```python
housing= strat_train_set.drop("median_house_value", axis=1)
housing_label=strat_train_set["median_house_value"].copy()
```

```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy="median")
housing_num = housing.select_dtypes(include=[np.number])
imputer.fit(housing_num)
```

```python
imputer.statistics_
```

```python
X = imputer.transform(housing_num)
```

```python
housing=pd.DataFrame(X, columns=housing_num.columns, index=housing_num.index)
```
